
### 1) Earth-skimming EffArea generation

Source: Snowmass paper Figure 18 https://arxiv.org/pdf/2203.08096 at 90%CL

(Updated Trinity Sensitivity estimates from https://arxiv.org/pdf/2509.18236 Figure 2 at 90%CL)

TAMBO cross-referenced against https://arxiv.org/abs/2507.08070 Figure 3 Aperture Plot 


Quoted sensitivity input:
$$
E^2\Phi_{\mathrm{sens}}(E)
\quad
[\mathrm{GeV\,cm^{-2}\,s^{-1}\,sr^{-1}}]
\quad
\text{at }90\%\ \mathrm{CL}.
$$

Effective area from sensitivity:
$$
A_{\mathrm{eff}}(E)
=
\frac{\mu_{90}\,E}
{T\,\Delta\Omega\,\left(E^2\Phi_{\mathrm{sens}}(E)\right)\,\ln 10},
\qquad
\mu_{90}=2.44,\ \Delta\Omega=4\pi.
$$

Exposure:
$$
T=10\times 365.25\times 24\times 3600\ \mathrm{s}
\quad
(\text{POEMMA adjusted from quoted }5\ \text{yr to }10\ \text{yr in code}).
$$


TAMBO aperture decomposition:
$$
(A\Omega)_e(E)=\left|(A\Omega)_{\mathrm{all}}(E)-(A\Omega)_\tau(E)\right|.
$$

TAMBO aperture-to-area conversion:
$$
A_\tau(E)=\frac{(A\Omega)_\tau(E)\,C}{\Delta\Omega_{\mathrm{T}}\,T_{\mathrm{T}}},
\qquad
A_e(E)=\frac{(A\Omega)_e(E)\,C}{\Delta\Omega_{\mathrm{T}}\,T_{\mathrm{T}}},
$$
$$
\Delta\Omega_{\mathrm{T}}=\frac{4\pi}{0.1},\qquad T_{\mathrm{T}}=1\ \mathrm{yr}.
$$

Glashow-related fraction:
$$
p_{\bar{\nu}_e}(E)=\frac{A_e(E)}{A_e(E)+A_\tau(E)}.
$$

Combined Earth-skimming channels:
$$
A_\mu^{\mathrm{Earth}}(E)=0,
$$
$$
A_\tau^{\mathrm{Earth}}(E)=
A_\tau^{\mathrm{Trinity}}(E)+A^{\mathrm{POEMMA}}(E)+A_\tau^{\mathrm{TAMBO}}(E)+A^{\mathrm{GRAND}}(E)+A^{\mathrm{Auger}}(E),
$$
$$
A_e^{\mathrm{Earth}}(E)=A_e^{\mathrm{TAMBO}}(E)+A_e^{\mathrm{Trinity}}(E).
$$

## Event-generation equations (`EventGeneration.py`)

We impose
$$
f_e + f_\mu + f_\tau = 1,
$$
with Asimov/source fractions $(f_e^0,f_\mu^0,f_\tau^0)$.

For coarse bin $i=[E_i^{\min},E_i^{\max}]$, define
$$
I_{\alpha,i}
=
\Omega T
\sum_j
A_{\alpha,j}\,\Phi(E_j)\,\Delta E_{ij},
$$
$$
\Delta E_{ij}
=
\max\!\left(0,\min(e_{j+1},E_i^{\max})-\max(e_j,E_i^{\min})\right),
\quad
\alpha\in\{e,\mu,\tau,\mathrm{NC}\}.
$$

Poisson log-term:
$$
\log \mathrm{Pois}(k\mid \lambda)=k\log\lambda-\lambda-\log\Gamma(k+1).
$$

### 1) Tau channel

Observed (Asimov):
$$
N^{\mathrm{obs}}_{i,\tau}=I_{\tau,i}f_\tau^0+I_{e,i}f_e^0,
\qquad
N^{\mathrm{obs}}_{i,G}=I_{e,i}f_e^0.
$$

Expected:
$$
N^{\mathrm{exp}}_{i,\tau}=I_{\tau,i}f_\tau+I_{e,i}f_e,
\qquad
N^{\mathrm{exp}}_{i,e}=I_{e,i}f_e.
$$

Glashow probability:
$$
p_{G,i}
=
\mathrm{clip}\!\left(
\frac{N^{\mathrm{exp}}_{i,e}}{N^{\mathrm{exp}}_{i,\tau}}\,p^{\mathrm{in}}_{G,i},
0,1
\right).
$$

Tau likelihood:
$$
\log\mathcal L_{\tau}
=
\sum_i \log\mathrm{Pois}\!\left(N^{\mathrm{obs}}_{i,\tau}\mid N^{\mathrm{exp}}_{i,\tau}\right)
+
\sum_i \log\mathrm{Binom}\!\left(N^{\mathrm{obs}}_{i,G}\mid N^{\mathrm{obs}}_{i,\tau},p_{G,i}\right).
$$

### 2) Gen2 channel, optical

Per-bin expectation:
$$
N^{\mathrm{exp}}_{i,\mathrm{Gen2}}
=
I_{e,i}f_e+I_{\mu,i}f_\mu+I_{\tau,i}f_\tau.
$$

Track/cascade form used in pipeline:
$$
N^{\mathrm{exp}}_{i,c}
=
I^{(c)}_{e,i}f_e+I^{(c)}_{\mu,i}f_\mu+I^{(c)}_{\tau,i}f_\tau,
\quad c\in\{\mathrm{track},\mathrm{casc}\}.
$$

Asimov counts:
$$
K_{i,c}=N^{\mathrm{exp}}_{i,c}(f_e^0,f_\mu^0,f_\tau^0).
$$

Likelihood:
$$
\log\mathcal L_{\mathrm{Gen2}}
=
\sum_i \log\mathrm{Pois}(K_{i,\mathrm{track}}\mid N^{\mathrm{exp}}_{i,\mathrm{track}})
+
\sum_i \log\mathrm{Pois}(K_{i,\mathrm{casc}}\mid N^{\mathrm{exp}}_{i,\mathrm{casc}}).
$$

### 3) Radio channel

Observed totals:
$$
N^{\mathrm{obs}}_{i,\mathrm{tot}}
=
I_{e,i}f_e^0+I_{\mu,i}f_\mu^0+I_{\tau,i}f_\tau^0+I_{\mathrm{NC},i}.
$$

Observed ML-tagged:
$$
N^{\mathrm{obs}}_{i,\mathrm{ML}}
=
I_{e,i}f_e^0\,T_i+\left(1-I_{e,i}f_e^0\right)\,0.02.
$$

Observed multi-shower:
$$
N^{\mathrm{obs}}_{i,\mathrm{mult}}
=
I_{\mu,i}f_\mu^0\,M_i+I_{\tau,i}f_\tau^0\,R_i.
$$

Expected:
$$
N^{\mathrm{exp}}_{i,\mathrm{tot}}
=
I_{e,i}f_e+I_{\mu,i}f_\mu+I_{\tau,i}f_\tau+I_{\mathrm{NC},i},
$$
$$
N^{\mathrm{exp}}_{i,e}=I_{e,i}f_e,\quad
N^{\mathrm{exp}}_{i,\mu}=I_{\mu,i}f_\mu,\quad
N^{\mathrm{exp}}_{i,\tau}=I_{\tau,i}f_\tau.
$$

Probabilities:
$$
p_{\mu,i}
=
\mathrm{clip}\!\left(
\frac{N^{\mathrm{exp}}_{i,\mu}}{N^{\mathrm{exp}}_{i,\mathrm{tot}}}r_{\mu,i},0,1
\right),
\qquad
p_{\tau,i}
=
\mathrm{clip}\!\left(
\frac{N^{\mathrm{exp}}_{i,\tau}}{N^{\mathrm{exp}}_{i,\mathrm{tot}}}r_{\tau,i},0,1
\right),
$$
$$
q_{\mathrm{TP},i}
=
\mathrm{clip}\!\left(
\frac{N^{\mathrm{exp}}_{i,e}}{N^{\mathrm{exp}}_{i,\mathrm{tot}}}T_i,0,1
\right),
\qquad
q_{\mathrm{FP},i}
=
\mathrm{clip}\!\left(
\left(1-\frac{N^{\mathrm{exp}}_{i,e}}{N^{\mathrm{exp}}_{i,\mathrm{tot}}}\right)F_i,0,1
\right).
$$

Radio likelihood:
$$
\log\mathcal L_{\mathrm{radio}}
=
\sum_i \log\mathrm{Pois}(N^{\mathrm{obs}}_{i,\mathrm{tot}}\mid N^{\mathrm{exp}}_{i,\mathrm{tot}})
+
\sum_i \log\!\left[
\sum_{k=k_0}^{k_1}
\mathrm{Binom}(k;n_i,p_{\mu,i})
\mathrm{Binom}(m_i-k;n_i,p_{\tau,i})
\right]
+
\sum_i \log\!\left[
\sum_{k=k_0'}^{k_1'}
\mathrm{Binom}(k;n_i,q_{\mathrm{TP},i})
\mathrm{Binom}(m_i'-k;n_i,q_{\mathrm{FP},i})
\right],
$$
with
$$
n_i=N^{\mathrm{obs}}_{i,\mathrm{tot}},\quad
m_i=N^{\mathrm{obs}}_{i,\mathrm{mult}},\quad
m_i'=N^{\mathrm{obs}}_{i,\mathrm{ML}},
$$
$$
k_0=\max(0,m_i-n_i),\quad
k_1=\min(m_i,n_i),\quad
k_0'=\max(0,m_i'-n_i),\quad
k_1'=\min(m_i',n_i).
$$


## Likelihood-ratio to CL (Plotting.py)



Likelihood-ratio definition:

$$
\ell(\theta)=\log \mathcal L(\theta),\qquad
\hat{\theta}=\arg\max_{\theta}\ell(\theta),\qquad
\lambda(\theta)=\frac{\mathcal L(\theta)}{\mathcal L(\hat{\theta})}.
$$

Quantity used in the plots:
$$
z(\theta)\equiv \ell(\theta)-\ell(\hat{\theta})=\log\lambda(\theta)\le 0.
$$

Wilks-equivalent test statistic:
$$
q(\theta)\equiv -2\log\lambda(\theta)
=2\!\left[\ell(\hat{\theta})-\ell(\theta)\right]
=-2\,z(\theta),
\qquad
q(\theta)\sim \chi^2_{2}.
$$

Confidence region at CL \(=\alpha\):
$$
\mathcal C_{\alpha}
=
\left\{\theta:\ q(\theta)\le \chi^2_{2,\alpha}\right\}
=
\left\{\theta:\ z(\theta)\ge -\frac{1}{2}\chi^2_{2,\alpha}\right\}.
$$

For 2 dof:
$$
\chi^2_{2,\alpha}=-2\ln(1-\alpha)
\quad\Longrightarrow\quad
z(\theta)\ge \ln(1-\alpha).
$$

Useful contour thresholds (equivalent forms):
$$
68\%:\ z\ge \ln(0.32)=-1.139 \ \Longleftrightarrow\ \Delta\chi^2\le 2.30
$$
$$
90\%:\ z\ge \ln(0.10)=-2.303 \ \Longleftrightarrow\ \Delta\chi^2\le 4.61
$$
$$
95\%:\ z\ge \ln(0.05)=-2.996 \ \Longleftrightarrow\ \Delta\chi^2\le 5.99
$$
$$
99\%:\ z\ge \ln(0.01)=-4.605 \ \Longleftrightarrow\ \Delta\chi^2\le 9.21
$$


## Muon Allowed Fractions (Figure3b.py)


$$
\text{Parameters: }\quad \mu \equiv f_\mu,\qquad \eta \equiv (f_e,f_\tau).
$$

Profile log-likelihood:
$$
\ell_{p,i}(\mu)\equiv \max_{\eta}\,\ell_i(\mu,\eta).
$$

Best-fit (profiled) point:
$$
\hat{\mu}_i \equiv \arg\max_{\mu}\ell_{p,i}(\mu),\qquad
\ell_{p,i}(\hat{\mu}_i)=\ell_{\max,i}.
$$

Profile likelihood ratio:
$$
\lambda_i(\mu)\equiv
\frac{\mathcal L_{p,i}(\mu)}{\mathcal L_{p,i}(\hat{\mu}_i)}
=
\frac{\mathcal L_i(\mu,\hat{\eta}_\mu)}{\mathcal L_i(\hat{\mu}_i,\hat{\eta})}.
$$

Test Statistic:
$$
z_i(\mu)\equiv \ell_{p,i}(\mu)-\ell_{p,i}(\hat{\mu}_i)
=\log\lambda_i(\mu)\le 0.
$$


PDF plotted:
$$
p_i(\mu)\propto e^{z_i(\mu)}
\quad\Rightarrow\quad
p_i(\mu)=\frac{e^{z_i(\mu)}}{\int e^{z_i(u)}\,du}.
$$
