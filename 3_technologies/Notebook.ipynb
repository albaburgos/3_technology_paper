Source Sensitivity: Snowmass paper Figure 18 https://arxiv.org/pdf/2203.08096 at 90%CL

Source Aperture: https://arxiv.org/pdf/2607.26128

Quoted sensitivity input:
$$
E^2\Phi_{\mathrm{sens}}(E)
\quad
[\mathrm{GeV\,cm^{-2}\,s^{-1}\,sr^{-1}}]
\quad
\text{at }90\%\ \mathrm{CL}.
$$

Effective area from sensitivity:
$$
A_{\mathrm{eff}}(E)
=
\frac{\mu_{90}\,E}
{T\,\Delta\Omega\,\left(E^2\Phi_{\mathrm{sens}}(E)\right)\,\ln 10},
\qquad
\mu_{90}=2.44,\ \Delta\Omega=4\pi.
$$

Exposure:
$$
T=10\times 365.25\times 24\times 3600\ \mathrm{s}
\quad
(\text{POEMMA adjusted from quoted }5\ \text{yr to }10\ \text{yr in code}).
$$


TAMBO aperture decomposition:
$$
(A\Omega)_e(E)=\left|(A\Omega)_{\mathrm{all}}(E)-(A\Omega)_\tau(E)\right|.
$$

Aperture-to-average-effective-area conversion (TAMBO & Trinity, `effective_area_from_aperture`):
$$
\bar A_\tau(E)=\frac{(A\Omega)_\tau(E)\,C\,\Delta\Omega_{\mathrm{FOV}}}{4\pi},
\qquad
\bar A_e(E)=\frac{(A\Omega)_e(E)\,C\,\Delta\Omega_{\mathrm{FOV}}}{4\pi},
$$

$$
\Delta\Omega_{\mathrm{FOV}}=
\begin{cases}
0.2806\ \mathrm{sr} & \text{TAMBO: FOV } 120^\circ\ (\mathrm{az})\times 30^\circ\ (\mathrm{zen})\\[2pt]
0.09126\ \mathrm{sr} & \text{Trinity: FOV } 5^\circ\ \text{below horizon}\times 60^\circ\ (\mathrm{az})
\end{cases}
$$

## Event-generation equations (`EventGeneration.py`)

We impose
$$
f_e + f_\mu + f_\tau = 1,
$$
with Asimov/source fractions $(f_e^0,f_\mu^0,f_\tau^0)$.

For coarse bin $i=[E_i^{\min},E_i^{\max}]$, define the flavor-weighted integral
$$
I_{\alpha,i}
=
\Omega T
\sum_j
A_{\alpha,j}\,\Phi(E_j)\,\Delta E_{ij},
\qquad
\Delta E_{ij}
=
\max\!\left(0,\min(e_{j+1},E_i^{\max})-\max(e_j,E_i^{\min})\right),
\quad
\alpha\in\{e,\mu,\tau,\mathrm{NC}\}.
$$

Poisson log-term:
$$
\log \mathrm{Pois}(k\mid \lambda)=k\log\lambda-\lambda-\log\Gamma(k+1).
$$

### 1) Tau / Earth-skimming channel

Observed (Asimov):
$$
N^{\mathrm{obs}}_{i,\tau}=I_{\tau,i}f_\tau^0+I_{e,i}f_e^0,
\qquad
N^{\mathrm{obs}}_{i,G}=I_{e,i}f_e^0\,p_{G,i}.
$$

Expected:
$$
N^{\mathrm{exp}}_{i,\tau}=I_{\tau,i}f_\tau+I_{e,i}f_e,
\qquad
\lambda_{G,i} = I_{e,i}f_e\,p_{G,i}.
$$

Tau likelihood:
$$
\log\mathcal L_{\tau}
=
\sum_i \log\mathrm{Pois}\!\left(N^{\mathrm{obs}}_{i,\tau}\mid N^{\mathrm{exp}}_{i,\tau}\right)
+
\sum_i \log\mathrm{Pois}\!\left(N^{\mathrm{obs}}_{i,G}\mid \lambda_{G,i}\right).
$$

### 2) IceCube-Gen2 optical (TOISE), 5 independent channels

Channels $c \in \{\text{cascades, double-cascades, unshadowed-tracks, shadowed-tracks, starting-tracks}\}$ each have per-flavor effective areas $A^{(c)}_{e}, A^{(c)}_{\mu}, A^{(c)}_{\tau}$ (summed over $\nu$ and $\bar\nu$).

Per-channel, per-bin expected counts:
$$
N^{\mathrm{exp}}_{i,c}(f_e,f_\mu)
=
I^{(c)}_{e,i}\,f_e
+I^{(c)}_{\mu,i}\,f_\mu
+I^{(c)}_{\tau,i}\,(1-f_e-f_\mu).
$$

Asimov:
$$
K_{i,c}=N^{\mathrm{exp}}_{i,c}(f_e^0,f_\mu^0).
$$

Each channel contributes an independent Poisson term; only $f_e$ and $f_\mu$ are free ($f_\tau$ is constrained):
$$
\log\mathcal L_{\mathrm{Gen2}}
=
\sum_{c}\sum_i \log\mathrm{Pois}\!\left(K_{i,c}\mid N^{\mathrm{exp}}_{i,c}(f_e,f_\mu)\right).
$$

### 3) Radio channel

Observed totals:
$$
N^{\mathrm{obs}}_{i,\mathrm{tot}}
=
I_{e,i}f_e^0+I_{\mu,i}f_\mu^0+I_{\tau,i}f_\tau^0+I_{\mathrm{NC},i}.
$$

Observed ML-tagged:
$$
N^{\mathrm{obs}}_{i,\mathrm{ML}}
=
I_{e,i}f_e^0\,T_i+\left(N^{\mathrm{obs}}_{i,\mathrm{tot}}-I_{e,i}f_e^0\right)\cdot 0.02.
$$

Observed multi-shower:
$$
N^{\mathrm{obs}}_{i,\mathrm{mult}}
=
I_{\mu,i}f_\mu^0\,M_i+I_{\tau,i}f_\tau^0\,R_i.
$$

Expected sub-channel rates:
$$
\lambda_{i,\mathrm{ML}}
= I_{e,i}f_e\,T_i + \bigl(N^{\mathrm{exp}}_{i,\mathrm{tot}}-I_{e,i}f_e\bigr)\cdot F_i,
\qquad
\lambda_{i,\mathrm{mult}}
= I_{\mu,i}f_\mu\,r_{\mu,i}+I_{\tau,i}f_\tau\,r_{\tau,i}.
$$

Radio likelihood (Poisson on each observable):
$$
\log\mathcal L_{\mathrm{radio}}
=
\sum_i \log\mathrm{Pois}(N^{\mathrm{obs}}_{i,\mathrm{tot}}\mid N^{\mathrm{exp}}_{i,\mathrm{tot}})
+
\sum_i \log\mathrm{Pois}(N^{\mathrm{obs}}_{i,\mathrm{ML}}\mid \lambda_{i,\mathrm{ML}})
+
\sum_i \log\mathrm{Pois}(N^{\mathrm{obs}}_{i,\mathrm{mult}}\mid \lambda_{i,\mathrm{mult}}).
$$



## Likelihood-ratio to CL (Plotting.py)



Likelihood-ratio definition:

$$
\ell(\theta)=\log \mathcal L(\theta),\qquad
\hat{\theta}=\arg\max_{\theta}\ell(\theta),\qquad
\lambda(\theta)=\frac{\mathcal L(\theta)}{\mathcal L(\hat{\theta})}.
$$

Quantity used in the plots:
$$
z(\theta)\equiv \ell(\theta)-\ell(\hat{\theta})=\log\lambda(\theta)\le 0.
$$

Wilks-equivalent test statistic:
$$
q(\theta)\equiv -2\log\lambda(\theta)
=2\!\left[\ell(\hat{\theta})-\ell(\theta)\right]
=-2\,z(\theta),
\qquad
q(\theta)\sim \chi^2_{2}.
$$

Confidence region at CL \(=\alpha\):
$$
\mathcal C_{\alpha}
=
\left\{\theta:\ q(\theta)\le \chi^2_{2,\alpha}\right\}
=
\left\{\theta:\ z(\theta)\ge -\frac{1}{2}\chi^2_{2,\alpha}\right\}.
$$

For 2 dof:
$$
\chi^2_{2,\alpha}=-2\ln(1-\alpha)
\quad\Longrightarrow\quad
z(\theta)\ge \ln(1-\alpha).
$$

Useful contour thresholds (equivalent forms):
$$
68\%:\ z\ge \ln(0.32)=-1.139 \ \Longleftrightarrow\ \Delta\chi^2\le 2.30
$$
$$
90\%:\ z\ge \ln(0.10)=-2.303 \ \Longleftrightarrow\ \Delta\chi^2\le 4.61
$$
$$
95\%:\ z\ge \ln(0.05)=-2.996 \ \Longleftrightarrow\ \Delta\chi^2\le 5.99
$$
$$
99\%:\ z\ge \ln(0.01)=-4.605 \ \Longleftrightarrow\ \Delta\chi^2\le 9.21
$$

## Muon Allowed Fractions (Figure3b.py)

**Constraint at source:** $f_\tau^s = 0$, so
$$
f_e^s = 1 - f_\mu^s, \qquad f_\mu^s \in [0,1].
$$

**Oscillation to Earth**:
$$
P_{\alpha\to\beta} = \sum_i |U_{\alpha i}|^2\,|U_{\beta i}|^2,
$$
$$
f_e^{\oplus}(f_\mu^s) = P_{ee}\,f_e^s + P_{\mu e}\,f_\mu^s,
\qquad
f_\mu^{\oplus}(f_\mu^s) = P_{e\mu}\,f_e^s + P_{\mu\mu}\,f_\mu^s,
\qquad
f_\tau^{\oplus} = 1 - f_e^{\oplus} - f_\mu^{\oplus}.
$$
Mixing matrix $U$ from NuFIT 5.2, normal ordering.

**Profile likelihood along the oscillation curve:**

$\ell(f_e, f_\mu)$ is computed on the 2-D Earth flavor grid and interpolated with `LinearNDInterpolator`. Restricting to the oscillation curve reduces to a 1-D profile:
$$
\tilde\ell(f_\mu^s) = \ell\!\left(f_e^{\oplus}(f_\mu^s),\, f_\mu^{\oplus}(f_\mu^s)\right),
\qquad
\hat{f}_\mu^s = \arg\max_{f_\mu^s}\,\tilde\ell(f_\mu^s).
$$

**Profile likelihood ratio plotted:**
$$
\lambda(f_\mu^s) = e^{\,\tilde\ell(f_\mu^s) - \tilde\ell_{\max}} \in [0,1].
$$

**Confidence intervals** via Wilks' theorem, 1 DOF ($f_\mu^s$ is the single parameter of interest):
$$
\mathrm{CI}_\alpha = \left\{f_\mu^s :\ -2\log\lambda(f_\mu^s) \leq \chi^2_{1,\alpha}\right\}
= \left\{f_\mu^s :\ \tilde\ell(f_\mu^s) \geq \tilde\ell_{\max} - \tfrac{1}{2}\chi^2_{1,\alpha}\right\}.
$$

| CL | $\chi^2_{1,\alpha}$ | $z$ threshold |
|----|---------------------|---------------|
| 68% | 1.00 | $-0.50$ |
| 95% | 3.84 | $-1.92$ |
